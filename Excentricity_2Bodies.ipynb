# Representación de la Solución Analítica del Problema 2 Cuerpos
## 9 de abril de 2026


```python
p, e = 1, 0.6

# Todo lo que termina en s son listas
tetaers = np.linspace(0, 2 * np.pi, 1000)
rs = p / (1 + e * np.cos(tetaers))


xfs  = rs * np.cos(tetaers)
yfs  = rs * np.sin(tetaers)
```

```python
plt.plot(xfs, yfs)
plt.plot(0, 0, 'yo', label='Foco')
plt.title('Órbita elíptica')
plt.plot(p * (1 - e), 0, 'ro', label='Periapsis')
plt.plot(p * (1 + e), 0, 'go', label='Apoapsis')
plt.axis('equal')
plt.grid()
plt.tight_layout()
plt.legend()
plt.show()
```

```python
from ipywidgets import interact, FloatSlider
```

```python
@interact(e= (0,1,0.1))
def dibuja_trayectoria(e=0.5):
    p = 1

    # Todo lo que termina en s son listas
    tetaers = np.linspace(0, 2 * np.pi, 1000)
    rs = p / (1 + e * np.cos(tetaers))


    xfs  = rs * np.cos(tetaers)
    yfs  = rs * np.sin(tetaers)


    plt.plot(xfs, yfs)

# Definimos manualmente el limite de los ejes para que no se mueva al cambiar la excentricidad
    plt.figuresize = (6, 6)

    plt.xlim(-2*p, 2*p)
    plt.ylim(-2*p, 2*p)

    plt.plot(0, 0, 'yo', label='Foco')
    plt.title('Órbita elíptica')
    plt.plot(p * (1 - e), 0, 'ro', label='Periapsis')
    plt.plot(p * (1 + e), 0, 'go', label='Apoapsis')
    plt.grid()
    plt.tight_layout()
    plt.legend()
    plt.show()

dibuja_trayectoria(0.7)
```


## Trayectoria de las Partículas con transformación de Jacobi

```python
m1, m2 = 1, 1
M = m1 + m2

rvs = np.array([xfs, yfs]).T

r1vs = m2 / M * rvs
r2vs = - m1 / M * rvs
```

```python
tetaers = np.linspace(0, 2 * np.pi, 1000)
rs = p / (1 + e * np.cos(tetaers))


xfs  = rs * np.cos(tetaers)
yfs  = rs * np.sin(tetaers)


plt.plot(r1vs[:,0], r1vs[:,1], label='Partícula 1')
plt.plot(r2vs[:,0], r2vs[:,1], label='Partícula 2')
plt.plot(0, 0, 'yo', label='Foco')
plt.title('Órbita elíptica')
plt.grid()
plt.tight_layout()
plt.legend()
plt.show()
```
